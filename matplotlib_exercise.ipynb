```python
# 使うライブラリ
# !pip install japanize-matplotlib

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import japanize_matplotlib
japanize_matplotlib.japanize()
```

## 1問目
使うcsvはhist_data.csvです。下にコードを書いてください。

![問1](./imgs/matplotlib/plot_1.png)

```python
# CSVファイルを読み込む
df = pd.read_csv('data/hist_data.csv')

# aデータとbデータを抽出
a_data = df['aデータ']
b_data = df['bデータ']

# ヒストグラムを作成
plt.figure(figsize=(10, 6))
plt.hist(a_data, bins=30, alpha=0.7, label='aデータ', color='yellow')
plt.hist(b_data, bins=30, alpha=0.5, label='bデータ', color='skyblue')

# グラフのタイトルと軸ラベルを設定
plt.title('aデータとbデータの温度に関する度数分布', fontsize=14)
plt.xlabel('温度(℃)', fontsize=12)
plt.ylabel('度数', fontsize=12)

# 凡例を表示
plt.legend()

# グラフを表示
plt.show()
```

## 2問目
使うcsvはscatter_data.csvです。下にコードを書いてください。

![問2](./imgs/matplotlib/plot_2.png)

```python
# CSVファイルを読み込む
df = pd.read_csv('data/scatter_data.csv')

# クラスごとに色を設定
colors = {0: 'red', 1: 'blue', 2: 'green'}

# 散布図を描く
plt.figure(figsize=(8, 6))
for class_label in df['class'].unique():
    subset = df[df['class'] == class_label]
    plt.scatter(subset['length (cm)'], subset['width (cm)'], 
                color=colors[class_label], label=f'class{class_label}')

# グラフのタイトルとラベルを設定
plt.title('クラス0～2のlengthとwidthに関する散布図')
plt.xlabel('length (cm)')
plt.ylabel('width (cm)')
plt.legend()
plt.grid(True)
# グラフを表示
plt.show()
```

## 3問目
使うcsvはline_data.csvです。下に2つの画像を表示するコードを書いてください。

![問3_1](./imgs/matplotlib/plot_3_1.png)

![問3_2](./imgs/matplotlib/plot_3_2.png)

```python
# CSVファイルを読み込む
df = pd.read_csv('data/line_data.csv')

# プロット領域を2つ作成
fig, axs = plt.subplots(2, 1, figsize=(8, 12))

# 1つ目のグラフ（件数の推移）
axs[0].plot(df['期間'], df['件数'], marker='o')
axs[0].set_title('期間ごとの件数の推移')
axs[0].set_xlabel('期間')
axs[0].set_xticklabels(df['期間'], rotation=30)
axs[0].set_ylabel('件数')
axs[0].grid(True)

# 2つ目のグラフ（費用の推移）
axs[1].bar(df['期間'], df['費用'])
axs[1].set_title('期間ごとの費用の推移')
axs[1].set_xlabel('期間')
axs[1].set_xticklabels(df['期間'], rotation=30)
axs[1].set_ylabel('費用 (円)')
axs[1].grid(True)

# グラフを表示
plt.tight_layout()
plt.show()
```
